# Local Watershed Variables Correlation

This section iterates through correlation CSVs for each watershed and water period. For each file it:
- Reads the correlation matrix CSV into a dataframe,
- Renames technical column names to more readable labels,
- Plots a heatmap of the full correlation matrix using `seaborn.heatmap`,
- Saves a high-resolution PNG to the same directory as the source CSV.

Run this section to regenerate PNG heatmaps for any updated correlation CSVs in `data_directory`.

```python
paths = sorted(glob.glob(os.path.join(data_directory, "*Watershed/corr_regional*.csv")) )
sel_paths = sorted([f for f in paths if ("corr_regional_" not in f)])
sel_paths
```

```python
for path in sel_paths:

    df = pd.read_csv(path)
    df = df.rename({"TSS_mean": "TSS", "area_km2": "Surface", "precipitation":"Precipitation", "mean_discharge":"Discharge",
                    "u_wind": "Eastward Wind","v_wind": "Northward Wind", "anthropogenic_km2": "Anthropic Area","Unnamed: 0": ""}, axis=1).set_index('')   
    df = df.rename({"TSS_mean": "TSS", "area_km2": "Surface", "precipitation":"Precipitation", "mean_discharge":"Discharge",
                    "u_wind": "Eastward Wind","v_wind": "Northward Wind", "anthropogenic_km2": "Anthropic Area"})
    filename = f'corr_{os.path.basename(path).split("_")[-1].split(".")[0]}.png'
    directory = os.path.dirname(path)
    print(f"Processing file: {filename}")
    print(df.head())
    plt.figure(figsize=(12,14))
    sns.heatmap(df, annot=True, cmap='coolwarm',cbar=False)
    plt.savefig(os.path.join(directory, filename),dpi=500)
    plt.show()
    
    plt.close()
```

# Dataframe of correlation results only for TSS

This section collects correlation coefficients and significance indicators specifically for TSS. For each water period it:
- Finds correlation CSVs matching the period and excludes confidence-interval files (if present),
- Extracts the `TSS_mean` column and renames columns to either `COEF` (coefficients) or `SIG` (significance),
- Maps raw variable names to reader-friendly labels,
- Concatenates the extracted TSS columns across files to produce a merged table, and
- Saves the merged table as `corr_TSS_<period>.csv` in the same directory.

Run this section after generating the correlation CSVs to produce consolidated TSS correlation reports.

```python
cor = sorted(glob.glob(os.path.join(data_directory, "*Watershed/corr_*.csv"))) 
cor
```

# REVISAR NOMES DE SAÍDA DAS TABELAS

```python
# Keep only files that DO NOT contain "pattern_to_exclude"
watersehds = ['regional', 'local']
water_period = ["LW","HW", "R", "F"]
for watershed in watersehds:
    cor_paths_w = sorted([f for f in cor if (watershed not in f)])
    cor_paths_w = sorted([f for f in cor_paths_w if ("ci" not in f)])

    for wp in water_period:
        cor_paths = sorted([f for f in cor_paths_w if (wp in f)])
        print(cor_paths)
        
        list_dfs = []

        for path in cor_paths:
            df_temp = pd.read_csv(path)
            
            df_temp = df_temp[['Unnamed: 0','TSS_mean']].copy()
            # print(df_temp.head())
            
            name_columns = f'{os.path.basename(path).split("_")[1]}'
            if "si" in name_columns:
                name_columns = "SIG"
                df_temp = df_temp.fillna(" ")
            if "csv" in name_columns:
                name_columns = "COEF"
            
            df_temp.columns = ["TSS Correlation", f'{name_columns.upper()}']
            df_temp = df_temp.loc[df_temp['TSS Correlation'] != "TSS_mean"].copy()
            rename_dict = {"area_km2": "Surface Area", "mean_chirps":"Precipitation", 
                    "mean_discharge":"Discharge", "u_wind": "Eastward Wind Speed",
                    "v_wind": "Northward Wind Speed", "anthropogenic_km2": "Anthropic Area"}
            df_temp['TSS Correlation'] = df_temp['TSS Correlation'].map(rename_dict)
            list_dfs.append(df_temp.set_index('TSS Correlation') )

        df_merged = pd.concat(list_dfs, axis=1)
        filename = f'corr_TSS_{os.path.basename(path).split("_")[-1].split(".")[0]}.csv'
        print(filename)
        directory = os.path.dirname(path)
        # print(directory)
        df_merged.to_csv(os.path.join(directory, filename))
        print(df_merged.head())
```
