# Transform into remote sensing  reflectance and sunglint correction

```python
# Convert surface reflectance to remote sensing reflectance and apply glint correction
# Steps:
# 1) Divide surface reflectance by pi to obtain R_rs (remote sensing reflectance)
# 2) Subtract the SWIR band from VNIR bands following Wang & Shi (2007) and Maciel et al. (2021):
#    R_rs(VNIR) = R_sat(VNIR)/pi - R_sat(SWIR)/pi  -> implemented as Rrs - SWIR
# The SWIR band is assumed to capture specular water-surface reflection and thus
# can be used to remove glint even in turbid waters.
def deglint(img):
    # divide radiance/reflectance by pi to get R_rs
    Rrs = img.divide(math.pi)
    # subtract SWIR (swir1) from VNIR bands (blue, green, red, nir)
    deglint = (Rrs.select(['blue','green','red','nir','swir1','swir2'])
               .subtract(Rrs.select('swir1')))

    # preserve important metadata properties for later filtering and export
    return (deglint.copyProperties(img, ['system:time_start', 'CLOUD_COVER', "system:index"]))
```

# Sample data points pixel values

filter images that fall within a 16-day window period from each field point date

```python
def imgs_points(collection):
    """Return a function that for each field feature finds images within ±16 days.

    For each field point we:
    - build a 16-day before/after window (can be adjusted)
    - filter the collection by date and location (feature.geometry())
    - attach `dif_date_point` property with the image-to-field-day difference
    - limit to a reasonable number of images per point to avoid large lists
    """
    def wrap(feat):
        date_point = ee.Date(feat.get('system:time_start'))
        # window start/end in yyyy-MM-dd format
        data1 = date_point.advance(-16, 'day').format('yyyy-MM-dd')
        data2 = date_point.advance(16, 'day').format('yyyy-MM-dd')

        # filter the collection by date & bounds, and attach temporal difference
        filtro = (collection.filterDate(data1, data2)
                 .filterBounds(feat.geometry())
                 .map(lambda img: img.set({
                     'dif_date_point': ee.Date(img.get('system:time_start')).difference(date_point, 'day')
                 }).copyProperties(feat, ['ID'])))

        # limit number of images returned per point to avoid overly large lists
        return ee.ImageCollection(filtro.limit(10))

    return wrap
```

obtain statistics for the same location of the field point - with a 3 pixel window - and filter out if more than 4 pixels are masked   

```python
prj = ld8.first().projection()
prj
```

```python
def sample_point(img):
    """Compute statistics in a 3x3 pixel window around the field point.

    Implementation details:
    - Find the matching feature by `ID`
    - Buffer the point by 45 meters and take bounding box (approx 3x3 at 30m)
    - Compute mean, median, min/max, stdDev, and count for VNIR bands
    - Return the feature with appended image-derived properties
    - A downstream filter will remove samples with fewer than 4 valid pixels
    """

    feat = roi_points.filter(ee.Filter.eq('ID', img.get('ID'))).first()
    # buffer ~45 m to capture a 3x3 pixel area for 30 m data
    geom = feat.geometry().buffer(45).bounds()

    mean = img.reduceRegion(geometry=geom,
                            scale=prj.nominalScale(),
                            crs=prj.crs(),
                            reducer=ee.Reducer.mean())
    median = img.reduceRegion(geometry=geom,
                              scale=prj.nominalScale(),
                              crs=prj.crs(),
                              reducer=ee.Reducer.median())
    minMax = img.reduceRegion(geometry=geom,
                              scale=prj.nominalScale(),
                              crs=prj.crs(),
                              reducer=ee.Reducer.minMax())
    count = img.reduceRegion(geometry=geom,
                              scale=prj.nominalScale(),
                              crs=prj.crs(),
                              reducer=ee.Reducer.count())
    std_dev = img.reduceRegion(geometry=geom,
                                scale=prj.nominalScale(),
                                crs=prj.crs(),
                                reducer=ee.Reducer.stdDev())

    # attach sampled statistics and metadata to the feature
    return feat.set({
        "system_index": img.get('system:index'),
        'CLOUD_COVER': img.get('CLOUD_COVER'),
        'img_date': ee.Date(img.get('system:time_start')).format(),
        'dif_date_point': img.get('dif_date_point'),

        'blue_mean': mean.get('blue'),
        'green_mean': mean.get('green'),
        'red_mean': mean.get('red'),
        'nir_mean': mean.get('nir'),

        'blue_median': median.get('blue'),
        'green_median': median.get('green'),
        'red_median': median.get('red'),
        'nir_median': median.get('nir'),

        'blue_min': minMax.get('blue_min'),
        'green_min': minMax.get('green_min'),
        'red_min': minMax.get('red_min'),
        'nir_min': minMax.get('nir_min'),

        'blue_max': minMax.get('blue_max'),
        'green_max': minMax.get('green_max'),
        'red_max': minMax.get('red_max'),
        'nir_max': minMax.get('nir_max'),

        'blue_stdDev': std_dev.get('blue'),
        'green_stdDev': std_dev.get('green'),
        'red_stdDev': std_dev.get('red'),
        'nir_stdDev': std_dev.get('nir'),

        "count_pixel": count.get('red')
    })

filter_count = ee.Filter.gt('count_pixel', 3)  # require at least 4 valid pixels (count > 3)
```
